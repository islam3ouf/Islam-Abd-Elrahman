```python

import numpy as np
import matplotlib.pyplot as plt

# ==========================
# Static Vector Visualization
# ==========================

plt.style.use("dark_background")

# Vector
v = np.array([4, 3])

# Figure
fig, ax = plt.subplots(figsize=(8, 8))
fig.patch.set_facecolor("#111827")
ax.set_facecolor("#111827")

# Limits
ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)

# Grid
ax.grid(True, linestyle='--', alpha=0.3)

# Axes
ax.axhline(0, color='white', linewidth=1)
ax.axvline(0, color='white', linewidth=1)

# Draw Vector
ax.quiver(
    0, 0,
    v[0], v[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='#00E5FF',
    linewidth=4
)

# Projection Lines
ax.plot([v[0], v[0]], [0, v[1]], '--', color='#FFB703')
ax.plot([0, v[0]], [v[1], v[1]], '--', color='#FFB703')

# Labels
ax.text(
    v[0] + 0.15,
    v[1] + 0.15,
    f"v = ({v[0]}, {v[1]})",
    fontsize=14,
    color='white',
    weight='bold'
)

# Magnitude
magnitude = np.linalg.norm(v)

ax.text(
    0.2,
    5.3,
    f"|v| = {magnitude:.2f}",
    fontsize=13,
    color='#00E5FF'
)

# Component Labels
ax.text(v[0]/2, -0.4, f"x = {v[0]}",
        color='#FFB703', fontsize=12)

ax.text(v[0]+0.2, v[1]/2, f"y = {v[1]}",
        color='#FFB703', fontsize=12)

# Title
ax.set_title(
    "Static Vector Visualization",
    fontsize=18,
    weight='bold',
    color='white',
    pad=20
)

# Axis Labels
ax.set_xlabel("X-axis", fontsize=12)
ax.set_ylabel("Y-axis", fontsize=12)

# Equal Scale
ax.set_aspect('equal')

plt.tight_layout()
plt.show()
```
