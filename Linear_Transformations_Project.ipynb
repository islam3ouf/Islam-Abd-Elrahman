````python
import numpy as np
import matplotlib.pyplot as plt

print("Libraries Loaded Successfully")
````

````python
square = np.array([
    [0, 0],
    [1, 0],
    [1, 1],
    [0, 1],
    [0, 0]
])

plt.figure(figsize=(6,6))
plt.plot(square[:,0], square[:,1], linewidth=3)

plt.xlim(-3,3)
plt.ylim(-3,3)
plt.grid(True)
plt.title("Original Shape")

plt.show()
````

````python
scaling_matrix = np.array([
    [2, 0],
    [0, 2]
])

scaled_square = square @ scaling_matrix.T

plt.figure(figsize=(6,6))

plt.plot(
    square[:,0],
    square[:,1],
    label="Original",
    linewidth=3
)

plt.plot(
    scaled_square[:,0],
    scaled_square[:,1],
    label="Scaled",
    linewidth=3
)

plt.xlim(-3,3)
plt.ylim(-3,3)

plt.grid(True)
plt.legend()
plt.title("Scaling Transformation")
plt.savefig("scaling.png")
plt.show()
````

````python
theta = np.radians(45)

rotation_matrix = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
])

rotated_square = square @ rotation_matrix.T

plt.figure(figsize=(6,6))

plt.plot(
    square[:,0],
    square[:,1],
    label="Original",
    linewidth=3
)

plt.plot(
    rotated_square[:,0],
    rotated_square[:,1],
    label="Rotated",
    linewidth=3
)

plt.xlim(-3,3)
plt.ylim(-3,3)

plt.grid(True)
plt.legend()
plt.title("Rotation Transformation")
plt.savefig("rotation.png")
plt.show()
````

````python
reflection_matrix = np.array([
    [1, 0],
    [0, -1]
])

reflected_square = square @ reflection_matrix.T

plt.figure(figsize=(6,6))

plt.plot(
    square[:,0],
    square[:,1],
    label="Original",
    linewidth=3
)

plt.plot(
    reflected_square[:,0],
    reflected_square[:,1],
    label="Reflected",
    linewidth=3
)

plt.xlim(-3,3)
plt.ylim(-3,3)

plt.grid(True)
plt.legend()
plt.title("Reflection Transformation")
plt.savefig("Reflection.png")
plt.show()
````

````python
shear_matrix = np.array([
    [1, 1],
    [0, 1]
])

sheared_square = square @ shear_matrix.T

plt.figure(figsize=(6,6))

plt.plot(
    square[:,0],
    square[:,1],
    label="Original",
    linewidth=3
)

plt.plot(
    sheared_square[:,0],
    sheared_square[:,1],
    label="Sheared",
    linewidth=3
)

plt.xlim(-3,3)
plt.ylim(-3,3)

plt.grid(True)
plt.legend()
plt.title("Shearing Transformation")
plt.savefig("Shearing.png")
plt.show()
````
